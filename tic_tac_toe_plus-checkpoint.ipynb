```python
import random
import time
from IPython.display import clear_output

print("--- BOMB TIC-TAC-TOE ---")

#CREATE A BLANK 5X5 BOARD
board = []
for i in range(5):
    row = [" ", " ", " ", " ", " "]
    board.append(row)

#PLACE A HIDDEN RANDOM BOMB
bomb_row = random.randint(0, 4)
bomb_col = random.randint(0, 4)
bomb_active = True

#REMOVAL TURN TRACKING 
o_can_remove = True
x_can_remove = True

current_player = "O"  
o_score = 0
x_score = 0
game_is_running = True

# MAIN GAME LOOP
while game_is_running:
    clear_output(wait=True) 
        
    print("\n  0   1   2   3   4")
    for r in range(5):
        print(r, end=" ")
        for c in range(5):
            print(board[r][c], end="")
            if c < 4:
                print(" | ", end="")
        print()
        if r < 4:
            print("  -----------------")
    print()

    print("Current Score -> O:", o_score, "| X:", x_score)
    print("Turn of player:", current_player)

#CHOOSE ACTION (Place a mark or Remove an opponent's mark)
    action = "place"
    
    has_removal = False
    if current_player == "O" and o_can_remove:
        has_removal = True
    else:
        if current_player == "X" and x_can_remove:
            has_removal = True
#player choose
    if current_player == "O" and has_removal:
        print("Choose: (1) Place a mark or (2) Remove opponent mark")
        choice= input("Enter 1 or 2: ")
        if choice == "2":
            action = "remove"
#computer chooses:
    else:
        if current_player == "X" and has_removal:
            o_cells = []
            for row_i in range(5):
                for col_i in range(5):
                    if board[row_i][col_i] == "O":
                        o_cells.append([row_i, col_i])
            if o_cells:
                if random.randint(1,10) <=3:
                    action = "remove"

    
    if action == "remove":
        opponent = "X"
        if current_player == "X":
            opponent = "O"
            
        r = int(input("Enter row to remove (0-4): "))
        c = int(input("Enter col to remove (0-4): "))
        
        if board[r][c] == opponent:
            board[r][c] = " "  
            print("Removed opponent mark!")
            
            
            if current_player == "O":
                o_can_remove = False
            else:
                x_can_remove = False
        else:
            print("No opponent mark there! Turn skipped.")
            time.sleep(2)
            
#Switch players
        if current_player == "O": 
            current_player = "X"
        else: 
            current_player = "O"
        continue

    if current_player == "O":
        r = int(input("Enter row to place (0-4): "))
        c = int(input("Enter col to place (0-4): "))
    else:
        
        empty_cells = []
        for row_i in range(5):
            for col_i in range(5):
                if board[row_i][col_i] == " ":
                    empty_cells.append([row_i, col_i])
        
        chosen_cell = random.choice(empty_cells)
        r = chosen_cell[0]
        c = chosen_cell[1]
        print("Computer (X) placed at:", r, c)
        time.sleep(1.5)

    
    if bomb_active and r == bomb_row and c == bomb_col:
        print(" BOOOOOOOM! Move cancelled!")
        bomb_active = False  
        time.sleep(2.5)
    else:
        board[r][c] = current_player  

#CHECK WIN CONDITION (4 consecutive matching marks)
        has_winner = False
        for row_i in range(5):
            for col_i in range(5):
                # Horizontal check
                if col_i <= 1:
                    if board[row_i][col_i] == current_player and board[row_i][col_i+1] == current_player and board[row_i][col_i+2] == current_player and board[row_i][col_i+3] == current_player:
                        has_winner = True
                # Vertical check
                if row_i <= 1:
                    if board[row_i][col_i] == current_player and board[row_i+1][col_i] == current_player and board[row_i+2][col_i] == current_player and board[row_i+3][col_i] == current_player:
                        has_winner = True
                # Down-right diagonal check
                if row_i <= 1 and col_i <= 1:
                    if board[row_i][col_i] == current_player and board[row_i+1][col_i+1] == current_player and board[row_i+2][col_i+2] == current_player and board[row_i+3][col_i+3] == current_player:
                        has_winner = True
                # Up-right diagonal check
                if row_i >= 3 and col_i <= 1:
                    if board[row_i][col_i] == current_player and board[row_i-1][col_i+1] == current_player and board[row_i-2][col_i+2] == current_player and board[row_i-3][col_i+3] == current_player:
                        has_winner = True

        if has_winner:
            clear_output(wait=True)
            print("\n  0   1   2   3   4")
            for row_final in range(5):
                print(row_final, end=" ")
                for col_final in range(5):
                    print(board[row_final][col_final], end="")
                    if col_final < 4: print(" | ", end="")
                print()
                if row_final < 4: print("  -----------------")
            
            print("\n PLAYER", current_player, "WINS THIS ROUND!")
            if current_player == "O":
                o_score = o_score + 1
            else:
                x_score = x_score + 1
            game_is_running = False
            break

#DRAW CONDITION:
    board_is_full = True
    for each_row in board:
        for cell in each_row:
            if cell == " ":
                board_is_full = False

    if board_is_full:
        print("DRAW! PLAY AGAIN!")
        time.sleep(3)
        board = []
        for i in range(5):
            row = [" ", " ", " ", " ", " "]
            board.append(row)
        bomb_row = random.randint(0, 4)
        bomb_col = random.randint(0, 4)
        bomb_active = True
        continue

#SWITCH PLAYERS FOR THE NEXT TURN
    if current_player == "O":
        current_player = "X"
    else:
        current_player = "O"
```
